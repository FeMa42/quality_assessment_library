## Set Paths

````python
input_folder = "./example_data/metrics_3d/Ground_Truth"
output_folder = "./example_data/metrics_3d/Ground_Truth_Preprocessed"
````

## Norm scaling & Centering
- Scales the longest axis of the model to 1
- Centers the object in the center of the frame
- Skips all files that are corrupted / not usable before or after cleaning the mesh

See Documentation for more configurations:

````python
from metrics_3d.preprocessing_3d import process_ground_truth_folder

results = process_ground_truth_folder(
    input_folder=input_folder,
    output_folder=output_folder,
    normalize_scale=True,
    center=True,
    logging=False,
    )
````

## OR

## Norm Scaling & Centering WITH Alignment

- same as above
- also aligns the model to the xy-plane with its shortes axis pointing towards z;  or however specified
- ⚠️ **WARNING:** automatic alignment can lead to unexpected results with different kinds of models -> always check the results of a couple examples visually

See Documentation for more configurations: [Documentation](./metrics_3d/README.md)

```python

````python
from metrics_3d.preprocessing_3d import process_ground_truth_folder

results = process_ground_truth_folder(
    input_folder=input_folder,
    output_folder=output_folder,
    alignment_method="xy_plane_shortest_axis",
    normalize_scale=True,
    center=True,
    logging=False,
    )
````

## Show Results

````python
import json
from pprint import pprint

# Pretty print results
pprint(results)

# Optionally, save to a JSON file as can be seen here:
with open("data_3d_preprocessing.json", "w") as f:
    json.dump(results, f, indent=2)
````
